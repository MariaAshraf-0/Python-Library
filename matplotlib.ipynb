```python
#you can also write 'markersize' = ms and 'markerfacecolor' = mfc
plt.plot(x,y1,marker='.',
         markersize=30,
         markerfacecolor='#1cd3fc',
         markeredgecolor='#1cd3fc',
        linestyle='solid', #dashed and dotted bhi likh sakty ho for no line we write 'none'
        linewidth=3,
        color='#1c5bfc'
        ) 
plt.plot(x,y2,marker = '.',
         markersize=30,
         markerfacecolor='#1cd3fc',
         markeredgecolor='#1cd3fc',
        linestyle='solid', #dashed and dotted bhi likh sakty ho for no line we write 'none'
        linewidth=3,
        color='#1c5bfc'
        )
```

```python
#-------plot x and y2 
plt.plot(x,y2,marker = '.',
         ms = 10,
        markerfacecolor='#1cd3fc',
        markeredgecolor='#1cd3fc',
        linestyle = 'dotted',
        linewidth=2,
        color='#1c7bfc'
        
        )
```

```python
#------------we also create dictipnary for parameter for line style 
```

```python
line_style= dict(
        marker = '.',
        markersize=30,
        markerfacecolor='#1cd3fc',
        markeredgecolor='#1cd3fc',
        linestyle='solid', #dashed and dotted bhi likh sakty ho for no line we write 'none'
        linewidth=3,
        color='#1c5bfc')
```

```python
#-----------------SAME RESULT------------
plt.plot(x,y1,**line_style)
plt.plot(x,y2,**line_style)
```

```python
y3=np.array([7,13,21,15])
```

```python
plt.plot(x,y1,**line_style)
plt.plot(x,y2,**line_style)
plt.plot(x,y3,**line_style)
```

```python
iris=pd.read_csv('iris.csv')
iris.sample(5)
```

```python
iris['Species']= iris['Species'].replace({'Iris-setosa':0,'Iris-virginica':1,'Iris-versicolor':2})
iris.sample(5)
```

```python
#cmap and c parameter (#cmap is used for diff color and c used for differentiate btw clases )
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal length')
```

```python
#also add color bar function
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal length')
plt.colorbar() 
```

```python
#alpha______
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.6)
plt.xlabel('Sepal Length')
plt.ylabel('Petal length')
plt.colorbar()
```

```python
#plot size  if you want to increase and decrease  graph size
plt.figure(figsize=(15,7)) #15 is width and 7 is height 

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=1)
plt.xlabel('Sepal Length')
plt.ylabel('Petal length')
plt.colorbar()
```
